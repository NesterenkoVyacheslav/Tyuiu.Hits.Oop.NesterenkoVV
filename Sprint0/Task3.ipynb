Описание задачи:
Создать базовый класс Event в C#, который будет представлять организованные
мероприятия. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Event:
• Атрибуты: ID мероприятия (EventId), Название мероприятия (EventName),
Дата проведения (Date).
• Методы:
o RegisterParticipant(): метод для регистрации участника.
o NotifyParticipants(): метод для уведомления участников о мероприятии.
o GetEventDetails(): метод для получения деталей мероприятия.
Требования к производным классам:
1. Конференция (Conference): Должна содержать дополнительные атрибуты,
такие как Количество спикеров (SpeakersCount).
Метод NotifyParticipants() должен быть переопределен для отправки
уведомлений с программой конференции.
2. Фестиваль (Festival): Должна содержать дополнительные атрибуты, такие как
Количество музыкальных групп (BandsCount).
Метод RegisterParticipant() должен быть переопределен для регистрации
участников с выбором интересующих музыкальных жанров.
3. Семинар (Seminar) (если требуется третий класс): Должна содержать
дополнительные атрибуты, такие как Тематические области (Topics).
Метод GetEventDetails() должен быть переопределен для отображения
тематических областей семинара вместе с другими деталями мероприятия.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Event
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }
    protected List<string> participants = new List<string>();

    public Event(int eventId, string eventName, DateTime date)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
    }

    public virtual void RegisterParticipant(string participantName)
    {
        participants.Add(participantName);
        Console.WriteLine($"Участник {participantName} зарегистрирован на мероприятие {EventName}");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление отправлено {participants.Count} участникам о мероприятии {EventName}");
    }

    public virtual string GetEventDetails()
    {
        return $"ID: {EventId}, Название: {EventName}, Дата: {Date:d}, Участников: {participants.Count}";
    }
}

public class Conference : Event
{
    public int SpeakersCount { get; set; }

    public Conference(int eventId, string eventName, DateTime date, int speakersCount)
        : base(eventId, eventName, date)
    {
        SpeakersCount = speakersCount;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление отправлено {participants.Count} участникам конференции {EventName}");
        Console.WriteLine($"Программа конференции включает {SpeakersCount} спикеров");
    }

    public override string GetEventDetails()
    {
        return $"ID: {EventId}, Название: {EventName}, Дата: {Date:d}, Участников: {participants.Count}, Количество спикеров: {SpeakersCount}";
    }
}

public class Festival : Event
{
    public int BandsCount { get; set; }
    private Dictionary<string, List<string>> participantGenres = new Dictionary<string, List<string>>();

    public Festival(int eventId, string eventName, DateTime date, int bandsCount)
        : base(eventId, eventName, date)
    {
        BandsCount = bandsCount;
    }

    public override void RegisterParticipant(string participantName)
    {
        participants.Add(participantName);
        participantGenres[participantName] = new List<string>();
        Console.WriteLine($"Участник {participantName} зарегистрирован на фестиваль {EventName}");
    }

    public void AddGenrePreference(string participantName, string genre)
    {
        if (participantGenres.ContainsKey(participantName))
        {
            participantGenres[participantName].Add(genre);
            Console.WriteLine($"Жанр '{genre}' добавлен для участника {participantName}");
        }
    }

    public override string GetEventDetails()
    {
        return $"ID: {EventId}, Название: {EventName}, Дата: {Date:d}, Участников: {participants.Count}, Количество групп: {BandsCount}";
    }
}

public class Seminar : Event
{
    public List<string> Topics { get; set; }

    public Seminar(int eventId, string eventName, DateTime date, List<string> topics)
        : base(eventId, eventName, date)
    {
        Topics = topics;
    }

    public override string GetEventDetails()
    {
        string topicsString = string.Join(", ", Topics);
        return $"ID: {EventId}, Название: {EventName}, Дата: {Date:d}, Участников: {participants.Count}, Тематические области: {topicsString}";
    }
}

Conference conference = new Conference(1, "IT Конференция 2026", new DateTime(2026, 10, 15), 5);
conference.RegisterParticipant("Иванов А.А.");
conference.RegisterParticipant("Петрова В.В.");
Console.WriteLine(conference.GetEventDetails());
conference.NotifyParticipants();

Console.WriteLine(new string('-', 40));

Festival festival = new Festival(2, "Рок Фестиваль", new DateTime(2026, 8, 20), 12);
festival.RegisterParticipant("Сидоров К.К.");
festival.AddGenrePreference("Сидоров К.К.", "Рок");
festival.AddGenrePreference("Сидоров К.К.", "Метал");
Console.WriteLine(festival.GetEventDetails());

Console.WriteLine(new string('-', 40));

List<string> seminarTopics = new List<string> { "Программирование", "Базы данных", "Веб-разработка" };
Seminar seminar = new Seminar(3, "Семинар по C#", new DateTime(2026, 9, 25), seminarTopics);
seminar.RegisterParticipant("Козлова М.М.");
Console.WriteLine(seminar.GetEventDetails());
seminar.NotifyParticipants();
```
